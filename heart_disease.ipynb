```python
#Task1
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
x_train ,x_test , y_train , y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
lr=LogisticRegression()
parameter_grid = {"penalty" : ["l1","l2",None],
              "C" : [0.05,0.5,1,2,3],
              "max_iter" :[100,200,300]}
log_reg_cv = GridSearchCV(LogisticRegression() , param_grid = parameter_grid, cv = 5)
log_reg_cv.fit(x_train_scaled, y_train)
```

```python
log_reg_cv.best_params_
```

```python
log_reg_cv.best_score_
```

```python
lr= LogisticRegression(C= 3 , max_iter = 100, penalty ='l2')
```

```python
lr.fit(x_train_scaled,y_train)
y_pred = lr.predict(x_test_scaled)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report, confusion_matrix
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True)
```

```python
#Task3
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
```

```python
#Task8
from sklearn.ensemble import StackingClassifier
```

```python
Estimators = np.array([LogisticRegression(), SVC(), SVC(kernel="poly", degree= 3),SVC(kernel="poly", degree= 5), DecisionTreeClassifier(), KNeighborsClassifier(n_neighbors = 3), KNeighborsClassifier(n_neighbors = 7)])
grid=[]
for i in range(5):
    alg = np.random.choice(Estimators, size = 3,replace= False)
    estimator = []
    for j in range(3):
        estimator.append((str(j), alg[j]))
    grid.append(estimator)
final = []
for i in range(5):
    alg = np.random.choice(Estimators, size = 1,replace= False)
    final.append(alg[0])
```

```python
grid
```

```python
final
```

```python
stack = GridSearchCV(estimator=StackingClassifier(estimators=[]), param_grid={"estimators": grid,"final_estimator":final}, cv=3)
```

```python
stack.fit(x_train_scaled, y_train)
print(stack.best_params_)
print(stack.best_score_)
```

```python
Stack=StackingClassifier(estimators=[('0', DecisionTreeClassifier()), ('1', LogisticRegression()), ('2', KNeighborsClassifier(n_neighbors=7))],
                           final_estimator= SVC())
```

```python
Stack.fit(x_train_scaled,y_train)
```

```python
y_pred = Stack.predict(x_test_scaled)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import classification_report, confusion_matrix
sns.heatmap(confusion_matrix(y_test,y_pred), annot=True)
```
